```python
import numpy as np
```

**1.** Given a random array, change the sign of elements whose values ​​are between 3 and 8

```python
arr = np.random.randint(0, 11, 10)
print(arr)
mask = (arr > 3) & (arr < 8)
arr[mask] *= -1
print(arr)
```

:**2.** Replace the maximum element of a random array with 0

```python
arr = np.random.randint(0, 11, 10)
print(arr)
index = np.argmax(arr)
arr[index] = 0
print(arr)
```

**3.** Construct the direct product of arrays (all combinations with each element). A two-dimensional array is provided as input.

```python
arr = np.array([
    [1, 2],
    [10, 20]
])

result = np.stack(np.meshgrid(arr[0], arr[1], indexing='ij'), axis=-1).reshape(-1, 2)

print(result)
```

**4.** Given two arrays, A (8x3) and B (2x2), find rows in A that contain elements from each row in B, regardless of the order of the elements in B.

```python
A = np.array([
    [1, 2, 9],
    [2, 1, 8],
    [3, 4, 7],
    [5, 4, 3],
    [1, 3, 6],
    [1, 4, 8],
    [2, 5, 6],
    [7, 8, 9]
])

B = np.array([
    [1, 2],
    [3, 4]
])

mask1 = np.isin(A, B[0])
row_mask1 = np.sum(mask1, axis=1) == 2
print(row_mask1)

mask2 = np.isin(A, B[1])
row_mask2 = np.sum(mask2, axis=1) == 2
print(row_mask2)

final_mask = row_mask1 | row_mask2

result = A[final_mask]

print(result)
```

**5.** Given a 10x3 matrix, find rows of unequal values ​​(for example, the row [2,2,3] remains, the row [3,3,3] is deleted)

```python
arr = np.array([
    [1, 2, 3],
    [2, 2, 3],
    [3, 3, 3],
    [4, 4, 4],
    [5, 5, 5],
    [6, 7, 6],
    [7, 8, 7],
    [8, 9, 8],
    [9, 10, 9],
    [10, 10, 10]
])

same = (arr[:, 0] == arr[:, 1]) & (arr[:, 1] == arr[:, 2])
print(same)
result = arr[~same]
print(result)
```

**6.** Given a two-dimensional array, remove the rows that are repeated.

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6],
    [1, 2, 3],
    [7, 8, 9],
    [4, 5, 6]
])

result = np.unique(arr, axis=0)
print(result)
```

* __Task 1__: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.  
 For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

```python
# Without numpy
X = [
    [1, 0, 1],
    [2, 0, 2],
    [3, 0, 3],
    [4, 4, 4]
]

result = 1

for i in range(min(len(X), len(X[0]))):
  if X[i][i] != 0:
    result *= X[i][i]

print(result)


# With numpy
X = np.array([
    [1, 0, 1],
    [2, 0, 2],
    [3, 0, 3],
    [4, 4, 4]
])
diagonal = np.diag(X)

result = np.prod(diagonal[diagonal != 0])
print(result)
```

* __Task 2__: Two vectors x and y are given. Check if they define the same multiset.  
  For example, for x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) answer is True.

```python
# Without numpy
x = [1, 2, 2, 4]
y = [4, 2, 1, 2]

sorted_x = sorted(x)
sorted_y = sorted(y)

if sorted_x == sorted_y:
  print(True)
else:
  print(False)


# With numpy
x = np.array([1, 2, 2, 4])
y = np.array([4, 2, 1, 2])

result = np.array_equal(np.sort(x), np.sort(y))
print(result)
```

* __Task 3__: Find the maximum element in the vector x among the elements preceded by zero.
 For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

```python
# Without numpy
x = [6, 2, 0, 3, 0, 0, 5, 7, 0]

values = []

for previous, current in zip(x[:-1], x[1:]):
  if previous == 0:
    values.append(current)

print(max(values))


# With numpy
x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])

result = np.max(x[1:][x[:-1] == 0])
print(result)
```

* __Task 4__: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.  
 For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
# Without numpy
x = [2, 2, 2, 3, 3, 3, 5]

values = [x[0]]
counts = [1]

for current in x[1:]:
  if current == values[-1]:
    counts[-1] += 1
  else:
    values.append(current)
    counts.append(1)

print((values, counts))


# With numpy
x = np.array([2, 2, 2, 3, 3, 3, 5])

unique, counts = np.unique(x, return_counts=True)
print((unique, counts))
```

* __Task 5__: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

```python
# Without numpy
import math

X = [[0, 0], [3, 4]]
Y = [[0, 0], [0, 4], [3, 0]]

result = [[math.dist(x, y) for y in Y] for x in X]

print(result)


# With numpy
X = np.array([[0, 0], [3, 4]])
Y = np.array([[0, 0], [0, 4], [3, 0]])

result = np.linalg.norm(X[:, None] - Y, axis=2)
print(result)
```
